```python
import pandas as pd
import plotly.express as px
```

```python
df = pd.read_csv("DataSet5-Safety-Enriched-Listings.csv")
```

```python
df = df.dropna(subset=['avg_safety_stars', 'avg_highway'])
```

```python
model_df = df.groupby(['make', 'model']).agg(
    avg_safety_stars=('avg_safety_stars', 'mean'),
    avg_highway=('avg_highway', 'mean'),
    avg_price=('price', 'mean')
).reset_index()
```

```python
bins = range(3000, 10000, 1000)
labels = [f'${i//1000}k-${(i+1000)//1000}k' for i in bins[:-1]]
model_df['price_bucket'] = pd.cut(model_df['avg_price'], bins=bins, labels=labels)
```

```python
model_df['car'] = model_df['make'] + ' ' + model_df['model']
```

```python
fig = px.scatter(
    model_df,
    x='avg_safety_stars',
    y='avg_highway',
    color='price_bucket',
    hover_name='car',
    hover_data={'avg_price': ':$.0f', 'price_bucket': False},
    title='Used Car Value: Safety vs Fuel Efficiency (1995-2009, Under $9,000)',
    labels={
        'avg_safety_stars': 'Average Safety Stars',
        'avg_highway': 'Average Highway MPG',
        'price_bucket': 'Price Range'
    }
)
```

```python
fig.show()
```

```python
fig.write_html("car_scatter_plot.html")
print("Saved car_scatter_plot.html")
```
