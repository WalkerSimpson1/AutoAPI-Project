```python
df = pd.read_csv("DataSet2-cleaned_listings.csv")
print(f"{df.shape[0]} rows and {df.shape[1]} columns in the dataset.")
```

```python
unique_cars = df[['make', 'model', 'year']].drop_duplicates().reset_index(drop=True)
print(f"Unique make/model/year combinations: {len(unique_cars)}")
```

```python
pd.set_option('display.max_rows', None)
print(df[['make', 'model']].drop_duplicates().sort_values(['make', 'model']).reset_index(drop=True))
```

```python
df = pd.read_csv("DataSet2-cleaned_listings.csv")
mpg = pd.read_csv("Claude-Listings-Matched-ToMPG.csv")

your_models = set(zip(df['make'], df['model']))
mpg_models = set(zip(mpg['make'], mpg['model']))

print("In cleaned listings but NOT in MPG file:")
for pair in sorted(your_models - mpg_models):
    print(pair)

print("\nIn MPG file but NOT in cleaned listings:")
for pair in sorted(mpg_models - your_models):
    print(pair)
```

```python
df = pd.read_csv("DataSet2-cleaned_listings.csv")
mpg = pd.read_csv("Claude-Listings-Matched-ToMPG.csv")

merged = df.merge(mpg, on=['make', 'model'], how='left')
print(merged.shape)
print(merged[['make', 'model', 'price', 'miles', 'avg_city', 'avg_highway', 'avg_combined']].head(10))
```

```python
merged.to_csv("DataSet4-MPGEnriched-Listings.csv", index=False)
print(f"Saved {len(merged)} listings to DataSet4-Enriched-Listings.csv")
```
